  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432059_MuhammadAlifHidayatullahHasib

# IMPORT LIBRARY
import os, re, json, copy, random, glob
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
from torch.nn.utils.rnn import pack_padded_sequence
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, classification_report, confusion_matrix
from IPython.display import display

# KONFIGURASI EKSPERIMEN
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
if DEVICE.type == "cpu":
    torch.set_num_threads(min(2, os.cpu_count() or 1))

print("PyTorch Version :", torch.__version__)
print("Device          :", DEVICE)
print("Random Seed     :", SEED)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432059_MuhammadAlifHidayatullahHasib

# 1. LOAD DAN VALIDASI DATASET
DATA_PATH = "SEntFiN-v1.1.csv"
if not os.path.exists(DATA_PATH):
    try:
        from google.colab import files
        DATA_PATH = next(iter(files.upload().keys()))
    except ImportError:
        pass
if not os.path.exists(DATA_PATH):
    raise FileNotFoundError("Dataset SEntFiN-v1.1.csv tidak ditemukan.")
df_raw = pd.read_csv(DATA_PATH, encoding="utf-8-sig")
missing = {"Title", "Decisions"} - set(df_raw.columns)
if missing:
    raise ValueError(f"Kolom tidak ditemukan: {missing}")
df_raw = df_raw.dropna(subset=["Title", "Decisions"]).copy()
print("Jumlah data mentah:", len(df_raw))
print("Jumlah kolom:", len(df_raw.columns))
display(df_raw.head())

# 2. PARSING ANOTASI JSON MENJADI PASANGAN HEADLINE-ENTITAS
CLASS_NAMES = ["negative", "neutral", "positive"]
LABEL_TO_ID = {label: i for i, label in enumerate(CLASS_NAMES)}
clean = lambda s: re.sub(r"\s+", " ", str(s)).strip()
rows = []
for _, row in df_raw.iterrows():
    decisions = json.loads(row["Decisions"])
    if not isinstance(decisions, dict):
        continue
    for entity, sentiment in decisions.items():
        sentiment = str(sentiment).strip().lower()
        if sentiment in LABEL_TO_ID:
            rows.append({"Title": clean(row["Title"]), "Entity": clean(entity), "Sentiment": sentiment})
annotations = pd.DataFrame(rows)
print("Jumlah anotasi sebelum cleaning:", len(annotations))

# 3. KUNCI NORMALISASI, DETEKSI KONFLIK, DAN DEDUPLIKASI
annotations["TitleKey"] = annotations["Title"].str.casefold()
annotations["EntityKey"] = annotations["Entity"].str.casefold()
label_counts = annotations.groupby(["TitleKey", "EntityKey"])["Sentiment"].nunique()
conflicting_titles = set(key[0] for key in label_counts[label_counts > 1].index)
annotations_clean = annotations[~annotations["TitleKey"].isin(conflicting_titles)].copy()
before_dedup = len(annotations_clean)
annotations_clean = annotations_clean.drop_duplicates(subset=["TitleKey", "EntityKey"]).reset_index(drop=True)
removed_duplicates = before_dedup - len(annotations_clean)

# 4. INPUT ENTITY-AWARE DAN LABEL ENCODING
annotations_clean["ModelText"] = annotations_clean.apply(lambda r: f"target {r['Entity']} sep headline {r['Title']}", axis=1)
annotations_clean["LabelID"] = annotations_clean["Sentiment"].map(LABEL_TO_ID)
assert not annotations_clean["LabelID"].isna().any()

# 5. PEMBAGIAN DATA BERDASARKAN HEADLINE UNIK (TANPA DATA LEAKAGE)
unique_titles = annotations_clean["TitleKey"].drop_duplicates().to_numpy()
train_titles, temp_titles = train_test_split(unique_titles, test_size=0.30, random_state=SEED, shuffle=True)
val_titles, test_titles = train_test_split(temp_titles, test_size=0.50, random_state=SEED, shuffle=True)
train_df = annotations_clean[annotations_clean["TitleKey"].isin(train_titles)].reset_index(drop=True)
val_df = annotations_clean[annotations_clean["TitleKey"].isin(val_titles)].reset_index(drop=True)
test_df = annotations_clean[annotations_clean["TitleKey"].isin(test_titles)].reset_index(drop=True)

# 6. TOKENISASI DAN VOCABULARY (HANYA DARI DATA TRAINING)
TOKEN_PATTERN = re.compile(r"[A-Za-z0-9]+(?:['’][A-Za-z]+)?")
def tokenize(text):
    return TOKEN_PATTERN.findall(str(text).lower())
MAX_VOCAB_SIZE, MAX_LEN = 20000, 40
token_counts = Counter(tok for text in train_df["ModelText"] for tok in tokenize(text))
word_to_index = {"<PAD>": 0, "<UNK>": 1}
for tok, _ in token_counts.most_common(MAX_VOCAB_SIZE - 2):
    word_to_index.setdefault(tok, len(word_to_index))

# 7. ENCODING DAN PADDING
def encode_dataframe(dataframe):
    sequences, lengths = [], []
    for text in dataframe["ModelText"]:
        ids = [word_to_index.get(t, 1) for t in tokenize(text)[:MAX_LEN]]
        lengths.append(max(1, len(ids)))
        sequences.append(ids + [0] * (MAX_LEN - len(ids)))
    labels = dataframe["LabelID"].astype(int).tolist()
    return torch.tensor(sequences, dtype=torch.long), torch.tensor(lengths, dtype=torch.long), torch.tensor(labels, dtype=torch.long)
X_train, len_train, y_train = encode_dataframe(train_df)
X_val, len_val, y_val = encode_dataframe(val_df)
X_test, len_test, y_test = encode_dataframe(test_df)

# 8. DATALOADER DAN BOBOT KELAS
BATCH_SIZE = 64
train_loader = DataLoader(TensorDataset(X_train, len_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(TensorDataset(X_val, len_val, y_val), batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(TensorDataset(X_test, len_test, y_test), batch_size=BATCH_SIZE, shuffle=False)
train_class_counts = torch.bincount(y_train, minlength=len(CLASS_NAMES)).float()
assert (train_class_counts > 0).all()
class_weights = len(y_train) / (len(CLASS_NAMES) * train_class_counts)

# 9. SIMPAN HASIL, RINGKASAN, DAN CEK DATA LEAKAGE
annotations_clean.drop(columns=["TitleKey", "EntityKey"]).to_csv("SEntFiN_v1.1_entity_pairs_cleaned.csv", index=False)
print("\n===== HASIL PREPROCESSING =====")
print("Data mentah              :", len(df_raw))
print("Anotasi awal             :", len(annotations))
print("Headline konflik         :", len(conflicting_titles))
print("Duplikasi pasangan dihapus:", removed_duplicates)
print("Pasangan akhir           :", len(annotations_clean))
print("Headline unik akhir      :", annotations_clean["TitleKey"].nunique())
print("Ukuran vocabulary         :", len(word_to_index))
print("Panjang sequence          :", MAX_LEN)
print("\n===== PEMBAGIAN DATA =====")
print("Training   :", len(train_df))
print("Validation :", len(val_df))
print("Testing    :", len(test_df))
no_leakage = (set(train_titles).isdisjoint(val_titles) and set(train_titles).isdisjoint(test_titles) and set(val_titles).isdisjoint(test_titles))
print("\nTidak ada headline tumpang tindih:", no_leakage)
assert no_leakage

# 10. EKSPLORASI DISTRIBUSI SENTIMEN
class_counts = annotations_clean["Sentiment"].value_counts().reindex(CLASS_NAMES, fill_value=0)
summary = pd.DataFrame({"Jumlah": class_counts, "Persentase (%)": (class_counts / class_counts.sum() * 100).round(2)})
display(summary)
plt.figure(figsize=(8, 5))
plt.bar(CLASS_NAMES, class_counts.values)
plt.title("Distribusi Kelas Sentimen SEntFiN v1.1")
plt.xlabel("Kategori Sentimen")
plt.ylabel("Jumlah Data")
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
display(annotations_clean[["Title", "Entity", "Sentiment", "ModelText"]].head(10))
```

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432059_MuhammadAlifHidayatullahHasib

# ARSITEKTUR MODEL LSTM
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim=100, hidden_size=64, num_classes=3, dropout_rate=0.30):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0)
        self.lstm = nn.LSTM(input_size=embedding_dim, hidden_size=hidden_size, num_layers=1, batch_first=True)
        self.dropout = nn.Dropout(dropout_rate)
        self.classifier = nn.Linear(hidden_size, num_classes)

    def forward(self, token_ids, sequence_lengths):
        embedded = self.embedding(token_ids)
        packed = pack_padded_sequence(embedded, sequence_lengths.cpu(), batch_first=True, enforce_sorted=False)
        _, (hidden_state, _) = self.lstm(packed)
        return self.classifier(self.dropout(hidden_state[-1]))

# INISIALISASI MODEL
EMBEDDING_DIM = 100
HIDDEN_SIZE = 64
DROPOUT_RATE = 0.30
model = SequenceModel(vocab_size=len(word_to_index), embedding_dim=EMBEDDING_DIM, hidden_size=HIDDEN_SIZE, num_classes=len(CLASS_NAMES), dropout_rate=DROPOUT_RATE).to(DEVICE)

print(model)
total_parameters = sum(p.numel() for p in model.parameters())
print("\nTotal parameter model:", total_parameters)
print("Device model:", next(model.parameters()).device)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432059_MuhammadAlifHidayatullahHasib

# 1. HYPERPARAMETER
EPOCHS = 20
LEARNING_RATE = 0.001
PATIENCE = 4
MAX_GRAD_NORM = 1.0

# 2. LOSS, OPTIMIZER, DAN PENYIMPAN RIWAYAT
criterion = nn.CrossEntropyLoss(weight=class_weights.to(DEVICE))
optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE)
history = {"train_loss": [], "val_loss": [], "train_accuracy": [], "val_accuracy": [], "val_macro_f1": []}
best_val_loss, best_epoch, epochs_without_improvement, best_model_state = float("inf"), 0, 0, None

def evaluate(loader):
    model.eval()
    loss_sum, y_true, y_pred, y_prob = 0.0, [], [], []
    with torch.no_grad():
        for tokens, lengths, labels in loader:
            logits = model(tokens.to(DEVICE), lengths)
            loss_sum += criterion(logits, labels.to(DEVICE)).item() * len(labels)
            probs = torch.softmax(logits, dim=1)
            y_true.extend(labels.tolist())
            y_pred.extend(probs.argmax(dim=1).cpu().tolist())
            y_prob.extend(probs.cpu().tolist())
    return loss_sum / len(loader.dataset), y_true, y_pred, y_prob

# 3. LOOP PELATIHAN
for epoch in range(1, EPOCHS + 1):
    model.train()
    train_loss_sum, train_true, train_pred = 0.0, [], []
    for tokens, lengths, labels in train_loader:
        tokens, labels_d = tokens.to(DEVICE), labels.to(DEVICE)
        optimizer.zero_grad()
        logits = model(tokens, lengths)
        loss = criterion(logits, labels_d)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), MAX_GRAD_NORM)
        optimizer.step()
        train_loss_sum += loss.item() * len(labels)
        train_true.extend(labels.tolist())
        train_pred.extend(logits.argmax(dim=1).detach().cpu().tolist())
    train_loss = train_loss_sum / len(train_loader.dataset)
    train_acc = accuracy_score(train_true, train_pred)
    val_loss, val_true, val_pred, _ = evaluate(val_loader)
    val_acc = accuracy_score(val_true, val_pred)
    val_f1 = precision_recall_fscore_support(val_true, val_pred, labels=[0, 1, 2], average="macro", zero_division=0)[2]
    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["train_accuracy"].append(train_acc)
    history["val_accuracy"].append(val_acc)
    history["val_macro_f1"].append(val_f1)
    print(f"Epoch {epoch:02d}/{EPOCHS} | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f} | Train Acc: {train_acc:.4f} | Val Acc: {val_acc:.4f} | Val Macro-F1: {val_f1:.4f}")
    if val_loss < best_val_loss - 1e-4:
        best_val_loss, best_epoch = val_loss, epoch
        best_model_state = copy.deepcopy(model.state_dict())
        epochs_without_improvement = 0
    else:
        epochs_without_improvement += 1
        if epochs_without_improvement >= PATIENCE:
            print("\nEarly stopping aktif.")
            break

# 4. MEMUAT DAN MENYIMPAN MODEL TERBAIK
assert best_model_state is not None
model.load_state_dict(best_model_state)
MODEL_PATH = "sentfin_lstm_model.pt"
torch.save({"model_state_dict": model.state_dict(), "word_to_index": word_to_index, "label_to_id": LABEL_TO_ID, "max_len": MAX_LEN, "embedding_dim": EMBEDDING_DIM, "hidden_size": HIDDEN_SIZE, "dropout_rate": DROPOUT_RATE}, MODEL_PATH)
print("\n===== TRAINING SELESAI =====")
print("Best Epoch:", best_epoch)
print("Best Validation Loss:", round(best_val_loss, 4))
print("Model tersimpan:", MODEL_PATH)
```
